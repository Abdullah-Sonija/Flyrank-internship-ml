## 2. The question: decision, action, cost of a wrong call

- **Decision:** Which pages should an editor review first for a refresh, expansion, or update?
- **Action:** A human editor reviews the top-ranked pages in the queue and decides whether to update the content, fix the meta data, or leave it alone.
- **Cost of a wrong call:** Wasted editor hours on a page that didn't need a refresh (false positive), or missing a high-value page that is actively decaying and losing traffic (false negative).
- **Why data/ML helps:** A simple rule (e.g., age > 180 and impressions > 1000) might miss nuanced decay patterns, prioritize pages that are already naturally recovering, or treat all positions equally. A ranking model can weigh multiple weak signals (position, age, competition, CTR) to create a more efficient queue than hardcoded if-statements.

## 3. Quick look at the data (2-3 real numbers)

The starter dataset contains 30,000 anonymized pages. Among these, there is a substantial number of pages with meaningful search visibility that are trending downwards. Specifically, there are 8,031 pages with more than 1,000 impressions in the last 90 days that show a "down" trend direction. This confirms there is a large, rich pool of candidates that require intelligent prioritization, making a scoring model highly relevant.

```python
import pandas as pd

# Load the starter dataset
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')

# Calculate the statistics
total_pages = len(df)
median_impressions = df['impressions_90d'].median()
high_imp_declining = len(df[(df['impressions_90d'] > 1000) & (df['trend_direction'] == 'down')])

print(f"Total pages in starter dataset: {total_pages:,}")
print(f"Median 90-day impressions: {median_impressions:,.0f}")
print(f"Pages with > 1,000 impressions and declining trend: {high_imp_declining:,}")
```
