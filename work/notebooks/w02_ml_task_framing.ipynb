## 1. My lane as an ML task (type)

**Type:** Ranking (via Classification)

Because the goal is to produce a prioritized queue for human editors who have limited capacity, the ultimate task is **Ranking**. We can achieve this by framing it as a **Classification** problem (e.g., predicting the probability of decline) and then sorting the candidates by that probability score.

```python
lane_type = 'Ranking / Classification'
print(f"Task Type: {lane_type}")
```

## 2. Target or proxy

**Target/Proxy:** Decline risk.

In the starter dataset, we use a proxy label: `is_declining_label` derived from the observed `trend_direction == 'down'`. This is based on an observed outcome (actual traffic dropping in the data window) rather than a subjective product rule. In a full warehouse environment, the true target would be an observed future decline in a forward-looking window (e.g., next 30 days) compared to the historical baseline.

```python
target = 'is_declining_label (trend_direction == "down")'
print(f"Target Proxy: {target}")
```

## 3. Success metric

**Metric:** Precision@K (e.g., Precision@50).

Because editors only have time to review a certain number of pages (capacity = K), global metrics like accuracy or ROC-AUC are less relevant than the quality of the top recommendations. 'Good' means that out of the top 50 pages the model recommends, a very high percentage are genuinely declining and worth a human's time.

```python
success_metric = 'Precision@K'
print(f"Metric: {success_metric}")
```

## 5. Why ML beats a fixed rule here

A fixed rule (e.g., `impressions_90d > 1000` AND `age_days > 180`) is brittle and treats all pages matching the criteria equally. The reality is messy: a page with 800 impressions, fierce competition, and rapidly dropping CTR is likely a more urgent fix than a stable 2,000-impression page gently aging. ML models (like a Random Forest) can discover these non-linear relationships and interactions between multiple weak signals (age, position, competition, word count) to produce a much more accurate prioritized queue than nested if-statements.

```python
print("ML discovers non-linear interactions between weak signals to prioritize candidates more effectively.")
```
